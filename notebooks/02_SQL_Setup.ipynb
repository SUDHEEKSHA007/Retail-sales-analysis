```python
from pathlib import Path
import pandas as pd
import sqlite3

project_root = Path.cwd()
if project_root.name.casefold() == "notebooks":
    project_root = project_root.parent
```

```python
cleaned_file = project_root / "data" / "cleaned" / "cleaned_superstore.csv"
if not cleaned_file.is_file():
    raise FileNotFoundError(
        f"Cleaned dataset not found: {cleaned_file}. Run "
        "01_Data_Cleaning.ipynb first."
    )

df = pd.read_csv(cleaned_file)
df.head()
```

```python
database_file = project_root / "sql" / "retail.db"
conn = sqlite3.connect(database_file)
```

```python
df.to_sql(
    "sales",
    conn,
    if_exists="replace",
    index=False
)

print("Database Created Successfully!")
```

```python
query = """
SELECT *
FROM sales
LIMIT 5;
"""

pd.read_sql(query, conn)
```

```python
conn.close()
```
